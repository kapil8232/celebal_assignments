## Part 5: Edge Case Handling

#### 1:What happens when order_items has an order_id not in orders?


```python
orders = pd.read_csv("cleaned_orders.csv")
order_items = pd.read_csv("order_items.csv")
invalid = order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]
invalid
```

#### 2:What happens when discount_percent > 100?


```python
order_items = pd.read_csv("order_items.csv")
df=order_items[order_items["discount_price"]>100]
df
```

#### 3:What happens when quantity is 0?


```python
items = pd.read_csv("order_items.csv")
df=items[items["quantity"] == 0]
df
```

#### 4:What happens when order_date is in the future?

```python
orders = pd.read_csv("cleaned_orders.csv")
orders["order_date"]=pd.to_datetime(orders["order_date"])
future = orders[orders["order_date"]>pd.Timestamp.today()]
future
```
