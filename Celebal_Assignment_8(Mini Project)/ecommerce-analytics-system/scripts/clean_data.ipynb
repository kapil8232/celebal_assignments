## Data cleaning

### Import required libraries

```python
import random
import pandas as pd
from faker import Faker
from datetime import datetime
```

### 1:Orders data cleaning

```python
df=pd.read_csv("orders.csv")
df.head(3)
```

```python
print(f"Shape:{df.shape}")
print(f"No of rows:{df.shape[0]}")
print(f"No of columns:{df.shape[1]}")
```

```python
print("No of null values in each column:")
df.isnull().sum()
```

```python
print("No of duplicate records:")
df[df.duplicated()].sum()
```

```python
def clean_orders(df):
     df["customer_id"]=df["customer_id"].fillna("unknown")
     df["order_date"]=pd.to_datetime(df["order_date"],format="mixed").dt.strftime("%Y-%m-%d %H:%M:%S")
     return df
df=clean_orders(df)
```

```python
print("After handling null values:")
df.isnull().sum()
```

```python
df.to_csv("cleaned_orders.csv", index=False)
```

### 2:Products data cleaning

```python
df=pd.read_csv("products.csv")
```

```python
print(f"Shape:{df.shape}")
print(f"No of rows:{df.shape[0]}")
print(f"No of columns:{df.shape[1]}")
```

```python
print("No of null values in each column:")
df.isnull().sum()
```

```python
print("No of duplicate records:")
df[df.duplicated()].sum()
```

```python
def clean_products(df):
    df["Product_name"]=df["Product_name"].str.strip()
    df["Product_name"]=df["Product_name"].str.title()
    return df
df=clean_products(df)
```

```python
df.to_csv("cleaned_products.csv" ,index=False)
```

### 3:Customers data cleaning

```python
df=pd.read_csv("customers.csv")
df.head(5)
```

```python
print(f"Shape:{df.shape}")
print(f"No of rows:{df.shape[0]}")
print(f"No of columns:{df.shape[1]}")
```

```python
print("No of null values in each column:")
df.isnull().sum()
```

```python
print("No of duplicate records:")
df[df.duplicated()].sum()
```

```python
def validate_emails(df):
    invalid_emails=df[
        ~df["email"].str.contains("@") |
        ~df["email"].str.contains(".com", regex=False)
]
    return invalid_emails[["customer_id", "email"]]
```

```python
validate_emails(df)
```

```python
df.to_csv("cleaned_customers.csv")
```

### 4:Invalid orders

```python
orders=pd.read_csv("orders.csv")
order_items=pd.read_csv("order_items.csv")
```

```python
def check_referential_integrity(order_items,orders):
    invalid_orders = order_items[
        ~order_items["order_id"].isin(orders["order_id"])
    ]
    return invalid_orders
    
df=check_referential_integrity(order_items,orders)
df
```

```python
order_items.to_csv("cleaned_order_items.csv")
```
