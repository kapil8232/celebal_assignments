# Data Cleaning

## 1:Import Libraries

```python
import random
import pandas as pd
from faker import Faker
from datetime import datetime
```

## 2:Data Generation

### 2.1 Customer_data

```python
numberof_customer=500
fake=Faker()
customer_types = ["REGULAR", "PREMIUM", "VIP"]
data=[]

for i in range(1,numberof_customer+1):
    if random.random()<.02:
        email=random.choice([
            fake.user_name()+"gmail.com",
            fake.user_name()+"@",
            fake.user_name()+"@gmail"
        ])
    else:
        email=fake.email()
    data.append({
        "customer_id":i,
        "customer_name":fake.name(),
        "email":email,
        "registration_date":fake.date_between("-5y","today"),
        "customer_type":random.choices(customer_types,weights=[70,20,10],k=1)[0]
    })

customer_df=pd.DataFrame(data)
customer_df.to_csv("customers.csv",index=False)
print("Data successfully saved")
```

### 2.2 Products Data

```python
categories = {
        "Electronics": ["Mobile", "Laptop", "Camera"],
        "Clothing": ["Men", "Women", "Kids"],
        "Home": ["Furniture", "Kitchen"],
        "Books": ["Comedy", "Education"]
}
products={
    "Mobile":["samsumgm31","oppo51","vivo32"],
    "Laptop":["macbook","ideapad","omen"],
    "Camera":["canon200","sony24","nikon300"],
    "Men":["jeans","shirt","lower"],
    "Women":["jeans","top","skirt"],
    "Kids":["jeans","pajama","lower"],
    "Furniture":["bed","table","chair"],
    "Kitchen":["box","dinning_set","mixer"],
    "Comedy":["omg","mice","tillu"],
    "Education":["cpp","alchemist","heaven"]
}
numberof_products=500
data=[]
for i in range(1,numberof_products+1):
    category=random.choice(list(categories.keys()))
    subcategory=random.choice(categories[category])
    product_name=random.choice(products[subcategory])

    if random.random()<.06:
        product_name=" "+product_name+" "
    data.append({
        "Product_id":i,
        "Product_name":product_name,
        "Category":category,
        "Subcategory":subcategory,
        "Cost_price":random.randint(1000,5000)
    })
product_df=pd.DataFrame(data)
product_df.to_csv("products.csv",index=False)
print("Data successfully saved")
```
