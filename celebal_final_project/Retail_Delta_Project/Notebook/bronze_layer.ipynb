## Step 1: Import Libraries

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import *
from pyspark.sql import SparkSession
from delta.tables import *
```

### Batch dataset path

```python
batch_path="/Volumes/dbacademy/default/data/retail_delta_project/datasets/batch"
```

## Step2:Defining function to write bronze delta tables

```python
def save_bronze_table(file_name,table_name):
    data=spark.read\
            .format("csv")\
            .option("header","true")\
            .option("inferSchema","true")\
                .load(f"{batch_path}/{file_name}")


    data=data.withColumn("ingesttime",current_timestamp())\
        .withColumn("source_file",col("_metadata.file_path"))\
            .withColumn("load_type",lit("batch"))
    
    data.write\
        .format("delta")\
        .mode("overwrite")\
            .saveAsTable(table_name)
    print(f"{table_name} loaded successfully")

    

    
        
```

## Step3:Function Calling 

```python
save_bronze_table( "orders_batch.csv","orders_bronze")

save_bronze_table("products_batch.csv","products_bronze")

save_bronze_table("customers_batch.csv","customers_bronze")

save_bronze_table("stores_batch.csv","stores_bronze")
```
