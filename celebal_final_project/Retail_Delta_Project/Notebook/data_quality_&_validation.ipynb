#1. Bronze vs Silver Count

```python
print("Bronze Orders:", spark.table("orders_bronze").count())
print("Silver Orders:", spark.table("silver_orders").count())

print("Bronze Customers:", spark.table("customers_bronze").count())
print("Silver Customers:", spark.table("silver_customers").count())

print("Bronze Products:", spark.table("products_bronze").count())
print("Silver Products:", spark.table("silver_products").count())

print("Bronze Stores:", spark.table("stores_bronze").count())
print("Silver Stores:", spark.table("silver_stores").count())
```

#2. Quarantine Records

```python
print("Orders Quarantine:",spark.table("orders_quarantine").count())

print("Customers Quarantine:", spark.table("customers_quarantine").count())

print("Products Quarantine:",spark.table("quarantine_products").count())
```

#3. SCD Validation

```python
print("Current Customers:", spark.table("dim_customers_scd2")\
      .filter("is_current=true")\
      .count())

print("Historical Customers:",spark.table("dim_customers_scd2")\
      .filter("is_current=false")\
      .count())
```

```python
print("Current Products:",spark.table("dim_products_scd2")\
      .filter("is_current=true")\
      .count())

print("Historical Products:", spark.table("dim_products_scd2")\
      .filter("is_current=false")
      .count())
```
