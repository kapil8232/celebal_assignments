#Sql Analysis

Query 1 : Total Revenue

```python
%sql
select round(sum(gross_amount),2) as total_revenue
from fact_orders
```

Query 2 : Total Orders

```python
%sql
select count(*) as total_orders
from fact_orders
```

Query 3 : Daily Sales

```python
%sql
select
date(order_ts) as order_date,
count(order_id) as total_orders,
round(sum(gross_amount),2) as total_sales
from fact_orders
group by date(order_ts)
order by order_date
limit 10;
```

Query 4 : Top 10 Customers by Revenue

```python
%sql
select customer_name,
round(sum(gross_amount),2) as total_revenue
from fact_orders fo
inner join
dim_customers_scd2 dc
on fo.customer_sk=dc.customer_sk
where dc.is_current="true"
group by dc.customer_name
order by total_revenue desc
limit 10; 
```

Query 5 : Top 10 Products

```python
%sql
select
p.product_name,
round(sum(f.gross_amount),2) as revenue
from fact_orders f
join dim_products_scd2 p
on f.product_sk = p.product_sk
where p.is_current = true
group by p.product_name
order by revenue desc
limit 10;
```

Query 6 : Category-wise Sales

```python
%sql
select
p.category,
round(sum(f.gross_amount),2) as revenue
from fact_orders f
join dim_products_scd2 p
on f.product_sk = p.product_sk
where p.is_current = true
group by p.category
order by revenue desc;
```

Query 7 : Region-wise Sales

```python
%sql
select
s.region,
round(sum(f.gross_amount),2) as revenue
from fact_orders f
join silver_stores s
on f.store_id = s.store_id
group by s.region
order by revenue desc;
```

Query 8 : Customer Segment Analysis

```python
%sql
select
c.segment,
count(f.order_id) as total_orders,
round(sum(f.gross_amount),2) as revenue
from fact_orders f
join dim_customers_scd2 c
on f.customer_sk = c.customer_sk
where c.is_current = true
group by c.segment
order by revenue desc;
```

Query 9 : Payment Method Analysis

```python
%sql
select
payment_method,
count(*) as total_orders,
round(sum(gross_amount), 2) as revenue
from fact_orders
group by payment_method
order by revenue desc;
```

Query 10 : Average Order Value

```python
%sql
select
ROUND(avg(gross_amount),2) as avg_order_value
from fact_orders;
```
