## Step4:Data Cleaning

#### 4.1:Create Quarantine Table

```python
orders_quarantine=bronze_orders.filter(
    col("customer_id").isNull()
    )
```

####4.2:Valid Records

```python
clean_orders=bronze_orders.filter(
    (~col("customer_id").isNull()) 
   )
```

####4.3 Remove Duplicates

```python
window=Window.partitionBy("order_id")\
    .orderBy(col("ingesttime").desc())
```

```python
clean_orders=clean_orders\
    .withColumn("rn",row_number().over(window))\
        .filter(col("rn")==1)\
            .drop("rn")
            
```

####4.4: Columns Data Type Casting

```python
clean_orders=clean_orders\
    .withColumn("order_ts",expr("try_cast(order_ts as timestamp)"), )\
        .withColumn("quantity",expr("try_cast(quantity as int)"))\
        .withColumn("unit_price", expr("try_cast(unit_price as double)"))\
        .withColumn("gross_amount",expr("try_cast(gross_amount as double)"))

```

#### 4.5:Filter invalid records

```python
invalid_record=clean_orders.filter(
    col("order_ts").isNull()|
    col("gross_amount").isNull()|
    col("quantity").isNull()|
    col("unit_price").isNull()
)
```

```python
orders_quarantine = orders_quarantine \
    .withColumn("order_ts",expr("try_cast(order_ts as timestamp)")) \
    .withColumn("quantity", expr("try_cast(quantity as int)")) \
    .withColumn("unit_price", expr("try_cast(unit_price as double)")) \
    .withColumn("gross_amount",expr("try_cast(gross_amount as double)"))
```

```python
orders_quarantine=orders_quarantine.unionByName(invalid_record)
```

#### 4.6:Valid Records

```python
clean_orders = clean_orders.filter(
    col("order_ts").isNotNull() &
    col("quantity").isNotNull() &
    col("unit_price").isNotNull() &
    col("gross_amount").isNotNull()
)
```

##Step5:Validation

```python
print("Bronze Orders:",bronze_orders.count())

print("Quarantine:",orders_quarantine.count())

print("Silver Orders:",clean_orders.count())
```

##Step 6 : Save Quarantine

```python
orders_quarantine.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("orders_quarantine")
```

## Step 7 : Save Silver Orders table

```python
clean_orders.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("silver_orders")
```
